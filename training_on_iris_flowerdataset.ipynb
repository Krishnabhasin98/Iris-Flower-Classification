# *CALCULATING AND ANALYZING THE ACCURACY OF IRIS FLOWER DATA SET*

## Import Libraries


```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset


```python
data = pd.read_csv("iris_main.csv")
```

## Data Visualization


```python
plt.figure(figsize=(8, 5))

sns.countplot(data=data, x='class')

plt.title("Distribution of Iris Flower Classes")
plt.xlabel("Flower Class")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(15,10))
plt.subplot(2,2,1)
sns.violinplot(x='class',y='Petal.Length',data=data)
plt.subplot(2,2,2)
sns.violinplot(x='class',y='Petal.Width',data=data)
plt.subplot(2,2,3)
sns.violinplot(x='class',y='Sepal.Length',data=data)
plt.subplot(2,2,4)
sns.violinplot(x='class',y='Sepal.Width',data=data)
```

## Prepare Features and Target


```python
y = data.iloc[:, -1]
x = data.iloc[:, :-1]
```

##### **SPLITTING THE DATASET INTO TRAIN AND TEST DATASET**

```python
train, test = train_test_split(data, test_size=0.2, random_state=42)
print(train.shape)
print(test.shape)
```

```python
test_x = test[['Sepal.Length', 'Sepal.Width', 'Petal.Length',
                 'Petal.Width']]
test_y = test['class']
train_x = train[['Sepal.Length', 'Sepal.Width', 'Petal.Length',
                 'Petal.Width']]
train_y = train['class']
```

##### ***USING RandomForestClassifier***

```python
model1 = RandomForestClassifier()
model1.fit(train_x.values,train_y.values)
prediction_1 = model1.predict(test_x.values)
print('Accuracy in %:',metrics.accuracy_score(test_y, prediction_1)*100)
```

```python
predictions_1 = model1.predict([[5.4, 3, 4.5, 1.5]])
```

```python
predictions_1
```

### Gaussian Naive Bayes


```python
model2 = GaussianNB()
model2.fit(train_x.values,train_y.values)
prediction_2 = model2.predict(test_x.values)
print('Accuracy in %:',metrics.accuracy_score(test_y,prediction_2)*100)
```

### Decision Tree Classifier


```python
model3 = DecisionTreeClassifier()
model3.fit(train_x.values,train_y.values)
prediction_3 = model3.predict(test_x.values)
print('Accuracy in %:',metrics.accuracy_score(test_y, prediction_3)*100)
```
